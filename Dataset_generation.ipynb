```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
from pandas import read_csv
from keras.models import Sequential
from keras.layers import Dense, Dropout, Activation, Flatten, LSTM, Input, BatchNormalization, ReLU, Bidirectional, GRU, LeakyReLU, Activation, Concatenate
from keras.models import Model
import tensorflow as tf

```

```python
df = pd.read_csv('~/Documents/notebook/gumi dataset/company-with-energy/dataset.csv')
```

```python
import os
import re

# Define the directory where you want to save the files
save_directory = "gumi-dataset-class"

# Ensure the directory exists (create it if it doesn't)
if not os.path.exists(save_directory):
    os.makedirs(save_directory)

# Function to sanitize the company names
def sanitize_filename(name):
    # Replace spaces, commas, and other invalid characters with underscores
    return re.sub(r'[\\/*?:"<>|, ]+', '_', name)

# Iterate over unique company values
for company in df['company'].unique():
    # Filter the data for the current company
    company_df = df[df['company'] == company]
    
    # Sanitize the company name for file name
    sanitized_company = sanitize_filename(company)
    
    # Define the file path to save the CSV
    file_name = f"{sanitized_company}_data.csv"
    full_path = os.path.join(save_directory, file_name)
    
    # Save the filtered data to a separate CSV file
    company_df.to_csv(full_path, index=False)
    
    print(f"Dataset for {company} saved as {full_path}")
```
